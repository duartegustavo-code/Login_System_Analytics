```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dados = pd.read_csv("dados_tratados.csv")
```

```python
total = len(dados)

# ano de pico
por_ano = dados.groupby("ANO_OCOR")["ID"].count().sort_index()
anos = por_ano.sort_values(ascending=False)
print(anos.head(1))
# macrorregião líder
macro = dados.groupby("MACRORREG_OCOR")["ID"].count().sort_values(ascending=False)
print(macro.head(1))
participacao = (macro.iloc[0] / total) * 100
# estado líder
ufs = dados.groupby("UF_OCOR")["ID"].count().sort_values(ascending=False)
print(ufs.head(1))
participacao_uf = (ufs.iloc[0] / total) * 100
# estado líder
meses = dados["MES_OCOR"].value_counts().sort_values(ascending=False)
print(meses.head(1))
```
